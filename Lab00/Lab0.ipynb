## Part 3: Linking Notebooks and Other Files

The cell below reads in a csv, a way for storing tabular data (csv = Comma-Separated Values).  When you run it, it will print several lines.  The first contains header information.  The next 5 contain data.  For this to work, the file `testcsv.csv` needs to be in the same directory as this notebook.

````python
import csv
with open("testcsv.csv", mode='r', newline='') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
````

**Task 4** Write a function `count_rows` that takes in a csv filepath (like `testcsv.csv` above) and returns the number of observations.  For the above, it should return 5, since there are 5 observations in the csv (the header row does not count).

````python
# BEGIN TASK 4
def count_rows(filepath):
    with open(filepath, mode='r', newline='') as file:
        reader = csv.reader(file)
        row_count = sum(1 for row in list(reader)[1:])
    return row_count
# END TASK 4

# Sanity check test
try:
    if count_rows("testcsv.csv")==5:
        print("Public test for task 4 passed")
    else:
        print("Public test for task 4 FAILED")
except:
    print("Public test for task 4 FAILED")
````
