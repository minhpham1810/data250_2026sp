1c: Because of the different scales for popularity scores (integers 0-100) and danceability scores (0-1), the above functionally sorts by popularity and then by danceability.  For instance, a song with popularity 90 and the wrost possible danceability (0) would be sorted above a song with popularity 89 and the highest possibility danceability (1).  For a more informative ranking, we should normalize the columns so that they are on the same scale.  

Create a new column `fun2` that stores $$\frac{popularity}{100} + danceability.$$  For instance,  if a song had danceability .5 and popularity 60, its `fun2` value would be  $\frac{60}{100}+.5=1.1$.   Then use `sort_by` (and one more pandas command) to show the top 10 songs with the highest `fun2` score.  

````python
# BEGIN QUESTION 1c
music['fun2'] = music['popularity']/100 + music['danceability']
music.sort_values('fun2', ascending=False).head(10)

# END QUESTION 1c
````

1d: Third, let's suppose we really want songs that are both especially highly popular and highly danceable.  E.g. we'd prefer a song with popularity 90 and danceability 0.9 (where both are high) to one with popularity 100 and danceability 0.85 (where we've compromised danceability for the sake of popularity). To do so, create a new column `fun3` that is the **minimum** of a songs dancability and (popularity/100). E.g. if a song had danceability .5 and popularity 60, its `fun3` value would be $min\{0.5, \frac{60}{100}\}=0.5.$ Then get the top 10 songs with the highest `fun3` score.  You might notice some minor differences.

````python
# BEGIN QUESTION 1d
music['fun3'] = music[['danceability', 'popularity']].apply(
    lambda row: min(row['danceability'], row['popularity']/100), axis=1)
music.sort_values('fun3', ascending=False).head(10)

# END QUESTION 1d
````

1e: Fourth, let's be inspired by regression where we squared residuals.  Create a new column `fun4` that is the sum of a songs danceability-squared and its (popularity/100)-squared. E.g. if a song had danceability .5 and popularity 60, its `fun4` value would be $0.5^2+(60/100)^2=0.61.$ Then get the top 10 songs with the highest `fun4` score.  

````python
# BEGIN QUESTION 1e
music['fun4'] = music[['danceability', 'popularity']].apply(
    lambda row: row['danceability']**2 + (row['popularity']/100)**2, axis=1
)
music.sort_values('fun4', ascending=False).head(10)
# END QUESTION 1e
````

1f: Finally, let's make what we just did even more extreme.  Create a new column `fun5` that is the sum of a songs danceability and its (popularity/100), after each is raised to the 50th power. E.g. if a song had danceability .5 and popularity 60, its `fun5` value would be $0.5^{50}+(60/100)^{50}.$ Then get the top 10 songs with the highest `fun5` score.   

````python
# BEGIN QUESTION 1f
music['fun5'] = music[['danceability', 'popularity']].apply(
    lambda row: row['danceability']**50 + (row['popularity']/100)**50, axis=1
)
music.sort_values('fun5', ascending=False).head(10)
# END QUESTION 1f
````

1g: Look at your work above.  Are all of these 5 approaches to finding highly popular and danceable songs distinct, or did any get the same top 10?  What do you notice about the approach from 1f specifically.  Can you think of another way to describe what it is effectively doing?  It might help to think about what happens when you take a small number (like .9) and raise it to the 50th power.

Looking at the five methods for identifying popular and danceable songs:

- fun1 is heavily dominated by popularity due to the scale difference (0-100 vs 0-1).
- fun2 normalizes the scales better, producing a more balanced ranking.
- fun3 prioritizes songs that excel in both dimensions equally.
- fun4 gives slightly more weight to extreme values.
- fun5 produces a dramatically different result. When you raise numbers between 0 and 1 to the 50th power, even values close to 1 become extremely small. This means fun5 is effectively filtering for songs where both danceability AND (popularity/100) are extremely close to 1, acting as a very strict filter that only rewards songs that are nearly perfect on both metrics simultaneously.
