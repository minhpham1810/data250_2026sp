## Question 1: Practice with Pandas Series

The following code creates a Pandas series with the index specified.   Imagine that this series encodes the number of squirels you counted on campus each month last Fall Semester.

1a. Use `.loc` notation to get the number of squirrels in October, storing it in `octloc`.  Then use `.iloc` notation to get the number of squirrels in October, storing it in `octiloc`.

## Question 4: Back to Baby Names

Lets return to the baby names data set from our code demos.  The code lodes this data set before.  **Please change `file_name` to be some state other than PA.  E.g. your home state (if not PA) or any other state that interests you.** 

```python
import pandas as pd # Convention for importing pandas
import numpy as np
import os.path
import zipfile

# No need to understand this code for now - but it opens pa.txt within a .zip folder
# and loads it as a data frame!
local_filename = "namesbystate.zip"
zf = zipfile.ZipFile(local_filename, 'r')

## MODIFY THIS TO BE FOR SOME STATE OTHER THAN PA
file_name = 'NY.TXT'

field_names = ['State', 'Sex', 'Year', 'Name', 'Count']
with zf.open(file_name) as fh:
    babynames = pd.read_csv(fh, header=None, names=field_names)

babynames.sample(10)
```

4a.  This question is a bit more open-ended.  Write code to answer the following questions below, and then briefly summarize your answers in the corresponding Markdown cell.  You will likely find it helpful to reference what we did in the code demos, and you might want to remember how we sorted by column in the `Working with an Extracted Column` column
- What was the most popular male name and the most popular female name in 1910? 
- What was the most popular male name and the most popular female name in your birth year? 
- What was the most popular male name and the most popular female name in 2024?
- In what year was your name the most popular?  If your name does not show up, that's OK - just note that.
- How many distinct names (across all years) begin with the first 3 letters of your first name?
- Are there any names that include all of a Y, Z, N, and D (upper or lowercase is OK)?
- When (if ever) was the name "Ace" first used?

```python
## Code for 4a here

# What was the most popular male name and the most popular female name in 1910? 
most_popular_male_name_1910 = babynames[(babynames['Year'] == 1910) & (babynames['Sex'] == 'M')].sort_values('Count', ascending=False).iloc[0]['Name']
print("Most popular male name in 1910: ",most_popular_male_name_1910)

most_popular_female_name_1910 = babynames[(babynames['Year'] == 1910) & (babynames['Sex'] == 'F')].sort_values('Count', ascending=False).iloc[0]['Name']
print("Most popular female name in 1910: ",most_popular_female_name_1910)

# What was the most popular male name and the most popular female name in your birth year? 
most_popular_male_name_2004 = babynames[(babynames['Year'] == 2004) & (babynames['Sex'] == 'M')].sort_values('Count', ascending=False).iloc[0]['Name']
print("Most popular male name in 2004: ",most_popular_male_name_2004)

most_popular_female_name_2004 = babynames[(babynames['Year'] == 2004) & (babynames['Sex'] == 'F')].sort_values('Count', ascending=False).iloc[0]['Name']
print("Most popular female name in 2004: ",most_popular_female_name_2004)

# What was the most popular male name and the most popular female name in 2024?
most_popular_male_name_2024 = babynames[(babynames['Year'] == 2024) & (babynames['Sex'] == 'M')].sort_values('Count', ascending=False).iloc[0]['Name']
print("Most popular male name in 2024: ",most_popular_male_name_2024)

most_popular_female_name_2024 = babynames[(babynames['Year'] == 2024) & (babynames['Sex'] == 'F')].sort_values('Count', ascending=False).iloc[0]['Name']
print("Most popular female name in 2024: ",most_popular_female_name_2024)

# In what year was your name the most popular?  If your name does not show up, that's OK - just note that.
most_popular_minh = babynames[babynames['Name'] == "Minh"].sort_values('Count', ascending=False).iloc[0]['Year']
print("The year when my name was most popular: ", most_popular_minh)

# How many distinct names (across all years) begin with the first 3 letters of your first name?
number_of_names_starting_with_min = babynames[babynames['Name'].str.startswith("Min")]['Name'].nunique()
print("Number of distinct names starting with 'Min': ", number_of_names_starting_with_min)

# Are there any names that include all of a Y, Z, N, and D (upper or lowercase is OK)?
letters = 'ynzdYNZD'
names_include_those_letters = len(babynames[babynames['Name'].str.contains(letters)])
print("Number of names containing letter Y, N, Z, D (both lowercase and uppercase: ", names_include_those_letters)

first_year_of_ace = babynames[babynames['Name'] == 'Ace']['Year'].min() if len(babynames[babynames['Name'] == 'Ace']) > 0 else "no Ace"
print("The year when the name Ace was first used: ",first_year_of_ace)
```
